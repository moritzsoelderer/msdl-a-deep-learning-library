### Neural Network Showcase
This noutebook showcases the training process of a neural network on the popular iris dataset with this custom library

```python
from engine.optimizer import SGD
import numpy as np
from model.layers import LinearLayer, DropOut
from model.neural_network import NeuralNetwork
from engine.activation import ReLU
from engine.loss import CrossEntropyWithLogits

np.random.seed(0)

nn = NeuralNetwork(layers=[
    LinearLayer(4, 32, ReLU),
    LinearLayer(32, 16, ReLU),
    DropOut(rate=0.2),
    LinearLayer(16, 3)
    ],
    loss=CrossEntropyWithLogits,
    optimizer=SGD(learning_rate=0.001, clip_gradients=1.0)
)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris

iris = load_iris()
X, y, class_names = iris.data, iris.target, iris.target_names

X_train, X_test, y_train, y_test = train_test_split(X, y)

encoder = OneHotEncoder(sparse_output=False)
y_train_cat = encoder.fit_transform(y_train.reshape(-1, 1))
y_test_cat = encoder.fit_transform(y_test.reshape(-1, 1))
```

```python
nn.train(X_train, y_train_cat, epochs=300, batch_size=24)
```

```python
predictions = nn.predict(X_test, y_test_cat)
y_pred = np.argmax(predictions, axis=1)
```

```python
from sklearn.metrics import accuracy_score

test_accuracy = accuracy_score(y_true=y_test, y_pred=y_pred)
f"Test Accuracy: {test_accuracy}"
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

disp = ConfusionMatrixDisplay.from_predictions(y_test, y_pred,
    display_labels=class_names,
    cmap=plt.cm.Blues,
    normalize='true'
)
disp.ax_.set_title("Confusion Matrix")
```
